## 准备：连上数据库（只需运行一次）

```python
import duckdb

con = duckdb.connect(r"C:\Users\user\Desktop\AI analysis\Data\aiusage.duckdb", read_only=True)
print("连上了")
```

## 第 1 条：看看表长什么样

`FROM sessions` = 从 sessions 表，`SELECT *` = 要所有列，`LIMIT 5` = 只要前 5 行

```python
con.sql("SELECT * FROM sessions LIMIT 5").df()
```

## 第 2 条：只要两列

把 `*` 换成列名，用逗号分隔

```python
con.sql("SELECT platform, user_turns FROM sessions LIMIT 5").df()
```

## 第 3 条：只要符合条件的行

`WHERE 条件` 筛选行。`=` 是判断相等，字符串用单引号

```python
con.sql("SELECT platform, user_turns FROM sessions WHERE platform = 'gemini' LIMIT 5").df()
```

## 第 4 条：数一数有多少行

`COUNT(*)` 数行数。结果应该是 1054

```python
con.sql("SELECT COUNT(*) FROM sessions").df()
```

## 第 5 条：数数 + 筛选

先筛出 gemini 的行再数。结果应该是 170

```python
con.sql("SELECT COUNT(*) FROM sessions WHERE platform = 'gemini'").df()
```

## 第 6 条：分组数数

`GROUP BY platform` 把行按平台分成几堆，`COUNT(*)` 对每堆分别数。

`AS sessions` 给这列起名，`ORDER BY ... DESC` 从大到小排序

```python
con.sql("""
    SELECT platform, COUNT(*) AS sessions
    FROM sessions
    GROUP BY platform
    ORDER BY sessions DESC
""").df()
```

上面最后一条用了三引号 `"""..."""`，这样 SQL 可以分成多行写，长查询更好读。
